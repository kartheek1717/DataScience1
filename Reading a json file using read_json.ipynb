```python
import json
import pandas as pd

# Step 1: Create sample data
students = [
    {
        "Name": "Kartheek",
        "Age": 20,
        "Course": "B.Tech",
        "Marks": 95
    },
    {
        "Name": "Rahul",
        "Age": 21,
        "Course": "B.Sc",
        "Marks": 88
    },
    {
        "Name": "Priya",
        "Age": 19,
        "Course": "BCA",
        "Marks": 92
    }
]

# Step 2: Write data to students.json
with open("students.json", "w") as file:
    json.dump(students, file, indent=4)

print("students.json created successfully!")

# Step 3: Read JSON file using pandas
df = pd.read_json("students.json")

# Step 4: Display the DataFrame
print("\nStudents Data:")
print(df)
```
