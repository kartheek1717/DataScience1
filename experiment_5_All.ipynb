```python
#Line plot
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [2, 4, 3, 5, 6]

plt.plot(x, y, marker='o')

plt.title("Line Plot")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")

plt.grid(True)
plt.savefig("line_plot.png")
plt.show()
```

```python
#Bar plot
import pandas as pd
import matplotlib.pyplot as plt

df = pd.DataFrame({'category': ['A', 'B', 'C'], 'value': [10, 20, 15]})
df.plot(kind='bar', x='category', y='value')
plt.title("Bar Plot")
plt.show()
```

```python
#Historygram and Destiny plot
import seaborn as sns
import numpy as np
data=np.random.randn(1000)
sns.histplot(data,kde=True)
plt.title("histogram and destiny")
plt.show()
```

```python
#Scatter plot
x=np.random.rand(50)
y=np.random.rand(50)
plt.scatter(x,y)
plt.title("scatter plot")
plt.xlabel("X")
plt.ylabel("Y")
plt.show()
```

```python
#box plot
sns.boxplot(data=np.random.randn(100,4))
plt.title("Box Plot")
plt.show()
```
