```python
import pandas as pd

data = {
    'Student': ['Akhil', 'Neha', 'Varun', 'Divya',
                'Rohan', 'Meena', 'Karthik', 'Sneha'],
    'Branch': ['CSE', 'ECE', 'CSE', 'EEE',
               'ECE', 'EEE', 'CSE', 'ECE'],
    'Year': [1, 2, 1, 3, 2, 3, 2, 1],
    'Marks': [72, 81, 89, 76, 93, 68, 84, 79]
}

df = pd.DataFrame(data)

print("Original DataFrame:")
print(df)

print("\n1. Group by Branch:")

group1 = df.groupby('Branch')

for name, group in group1:
    print("\nBranch:", name)
    print(group)

print("\nAverage Marks by Branch:")
print(df.groupby('Branch')['Marks'].mean())

print("\n2. Group by Branch and Year:")

group2 = df.groupby(['Branch', 'Year'])

for name, group in group2:
    print("\nGroup:", name)
    print(group)

print("\nAverage Marks by Branch and Year:")
print(df.groupby(['Branch', 'Year'])['Marks'].mean())
```

```python
import pandas as pd

data = {
    'Branch': ['CSE', 'CSE', 'CSE',
               'ECE', 'ECE', 'ECE',
               'EEE', 'EEE', 'EEE'],

    'Marks': [88, 74, 91,
              79, 86, 72,
              95, 83, 89],

    'Attendance': [92, 87, 96,
                   84, 91, 78,
                   97, 89, 94]
}

df = pd.DataFrame(data)

print("Original DataFrame:")
print(df)

result = df.groupby('Branch').aggregate({
    'Marks': ['sum', 'mean', 'std'],
    'Attendance': ['sum', 'mean', 'std']
})

print("\nSummary Statistics:")
print(result)
```

```python
import pandas as pd

url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/titanic.csv"

df = pd.read_csv(url)

print("First 5 Records:")
print(df.head())

print("\nDataset Information:")
print(df.info())

print("\nStatistical Summary:")
print(df.describe())

print("\n1. GroupBy using a Single Column:")

group_single = df.groupby('class')['fare'].agg(
    ['count', 'sum', 'mean', 'std', 'min', 'max']
)

print(group_single)

print("\n2. GroupBy using Multiple Columns:")

group_multiple = df.groupby(
    ['sex', 'class']
)['fare'].mean()

print(group_multiple)

print("\n3. Summary Statistics:")

summary = df.groupby('class').agg({
    'fare': ['mean', 'std', 'min', 'max'],
    'age': ['mean', 'std', 'min', 'max'],
    'sibsp': ['mean', 'max'],
    'parch': ['mean', 'max']
})

print(summary)

print("\n4. Number of Passengers in Each Class:")
print(df['class'].value_counts())

print("\nAverage Fare by Class:")
print(df.groupby('class')['fare'].mean())

print("\nAverage Age by Class:")
print(df.groupby('class')['age'].mean())

print("\nMaximum Fare by Class:")
print(df.groupby('class')['fare'].max())
```
