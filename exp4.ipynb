```python
import pandas as pd

dates = pd.to_datetime([
    '2025-05-10',
    '2025-05-11',
    '2025-05-12',
    '2025-05-13',
    '2025-05-14'
])

temperature = [25, 27, 26, 29, 30]

time_series = pd.Series(
    temperature,
    index=dates
)

print("Time Series:")
print(time_series)
```

```python
import pandas as pd

date_index = pd.date_range(
    start='2025-05-10',
    periods=5,
    freq='D'
)

print("Generated DatetimeIndex:")
print(date_index)
```

```python
import pandas as pd

india_dates = pd.date_range(
    start='2025-06-15 08:00',
    periods=4,
    freq='h',
    tz='Asia/Kolkata'
)

print("India Date Range:")
print(india_dates)

dates = pd.date_range(
    start='2025-06-15 08:00',
    periods=4,
    freq='h'
)

print("\nTimezone-naive Date Range:")
print(dates)

localized_dates = dates.tz_localize('Asia/Kolkata')

print("\nLocalized Dates:")
print(localized_dates)

new_york_dates = localized_dates.tz_convert(
    'America/New_York'
)

print("\nNew York Dates:")
print(new_york_dates)

india_series = pd.Series(
    [150, 250, 350, 450],
    index=localized_dates
)

new_york_series = pd.Series(
    [550, 650, 750, 850],
    index=new_york_dates
)

print("\nIndia Time Series:")
print(india_series)

print("\nNew York Time Series:")
print(new_york_series)

combined_series = pd.concat([
    india_series,
    new_york_series
])

print("\nCombined Time Series:")
print(combined_series)
```

```python
import pandas as pd

p = pd.Period('2025-05', freq='M')

print("Original Period:")
print(p)

print("\nAfter adding 2:")
print(p + 2)

print("\nAfter adding 4:")
print(p + 4)

print("\nAfter subtracting 1:")
print(p - 1)

print("\nAfter subtracting 3:")
print(p - 3)

periods = pd.period_range(
    start='2025-05',
    periods=5,
    freq='M'
)

print("\nRange of Periods:")
print(periods)
```

```python
import pandas as pd

p = pd.Period('2025-05', freq='M')

print("Original Period:")
print(p)

daily_start = p.asfreq(
    'D',
    how='start'
)

print("\nConverted to Daily - Start:")
print(daily_start)

daily_end = p.asfreq(
    'D',
    how='end'
)

print("\nConverted to Daily - End:")
print(daily_end)

period_index = pd.period_range(
    start='2025-05',
    periods=4,
    freq='M'
)

print("\nOriginal PeriodIndex:")
print(period_index)

daily_period_index = period_index.asfreq(
    'D',
    how='start'
)

print("\nPeriodIndex - Daily Start:")
print(daily_period_index)

daily_period_index_end = period_index.asfreq(
    'D',
    how='end'
)

print("\nPeriodIndex - Daily End:")
print(daily_period_index_end)
```
