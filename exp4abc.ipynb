```python
import pandas as pd
index=[
    ['Engineering','Engineering','mathematics','science'],
    ['AIML','DS','stastics','chemistry']
    ]
multi_index=pd.MultiIndex.from_arrays(index,names=['Department','branch'])
marks=pd.Series([98,97,99,100],index=multi_index)
print("Original series:")
print(marks)
print("\nData for Engineering:")
print(marks.loc['Engineering'])
print("\nData for AIML:")
print(marks.loc[('Engineering','AIML')])
print("\nData for science:")
print(marks.loc['science'])

```

```python
import pandas as pd

index = pd.MultiIndex.from_tuples(
    [
        ('Engineering', 'AIML'),
        ('Engineering', 'DS'),
        ('mathematics', 'satistics'),
        ('Science', 'Chemistry')
    ],
    names=['Department', 'Branch']
)

data = pd.DataFrame(
    {
        '2024': [78, 85, 98, 88],
        '2025': [90, 82, 95, 70]
    },
    index=index
)

print("Original Tabular Data:")
print(data)

unstacked_data = data.unstack()

print("\nData after Unstack():")
print(unstacked_data)

stacked_data = unstacked_data.stack(future_stack=True)

print("\nData after Stack():")
print(stacked_data)
```

```python
import pandas as pd
df1=pd.DataFrame(
    {
       'Name':['gani','bala','kushu'],
       'marks':[85,None,78],
       'Grade':['a','c','b']
    },
    index=[101,102,103]
)
df2=pd.DataFrame(
    {
        'Name':['shasi','vishnu','kani'],
        'marks':[99,100,98],
        'garde':['a','a','a']

    },
    index=[101,102,103]
)
print("First dataframe:")
print(df1)
print("Second dataframe:")
print(df2)
merged=pd.merge(
    df1,df2,
    left_index=True,
    right_index=True,
    how='outer',
    suffixes=('_df1','_df2')
)
print("\nMerged dataframe:")
print(merged)
combined=df1.combine_first(df2)
print("\n Combined dataframe:")
print(combined)

```
