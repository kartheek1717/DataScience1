```python
import pandas as pd

index = [
    ['Technology', 'Technology', 'Management', 'Management'],
    ['AI', 'Data Science', 'Finance', 'Marketing']
]


multi_index = pd.MultiIndex.from_arrays(
    index,
    names=['Department', 'Specialization']
)


marks = pd.Series(
    [91, 84, 76, 89],
    index=multi_index
)

print("Original Series:")
print(marks)


print("\nData for Technology:")
print(marks.loc['Technology'])

print("\nData for AI:")
print(marks.loc[('Technology', 'AI')])


print("\nData for Management:")
print(marks.loc['management'])
```

```python
import pandas as pd


index = pd.MultiIndex.from_tuples(
    [
        ('Technology', 'AI'),
        ('Technology', 'Data Science'),
        ('Management', 'Finance'),
        ('Management', 'Marketing')
    ],
    names=['Department', 'Specialization']
)


data = pd.DataFrame(
    {
        '2024': [88, 79, 91, 85],
        '2025': [94, 87, 96, 90]
    },
    index=index
)

print("Original Tabular Data:")
print(data)

unstacked_data = data.unstack()

print("\nData after Unstack():")
print(unstacked_data)

stacked_data = unstacked_data.stack()

print("\nData after Stack():")
print(stacked_data)
```

```python
import pandas as pd

# First DataFrame
df1 = pd.DataFrame(
    {
        'Name': ['Rahul', 'Priya', 'Vikram'],
        'Marks': [72, None, 86],
        'Grade': ['B', 'A', None]
    },
    index=[201, 202, 203]
)

# Second DataFrame
df2 = pd.DataFrame(
    {
        'Name': ['Rahul', 'Priya', 'Anjali'],
        'Marks': [78, 93, 81],
        'Grade': [None, 'A', 'B']
    },
    index=[201, 202, 204]
)

print("First DataFrame:")
print(df1)

print("\nSecond DataFrame:")
print(df2)

# Merge using index
merged = pd.merge(
    df1,
    df2,
    left_index=True,
    right_index=True,
    how='outer',
    suffixes=('_DF1', '_DF2')
)

print("\nMerged DataFrame:")
print(merged)

# Combine missing values
combined = df1.combine_first(df2)

print("\nDataFrame after combine_first():")
print(combined)
```
